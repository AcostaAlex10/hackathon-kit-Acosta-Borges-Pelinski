## Preprocesamiento mínimo

- Descartamos columnas "claramente" no informativas (IDs, códigos).
- Imputamos numéricas con la mediana y categóricas con la moda.
- Hacemos *label encoding* simple de las categóricas.

**Importante**: usamos las estadísticas del train para transformar el test (evita *leakage*).

```python
TARGET = 'nivel_urgencia'
ID_COL = 'id_paciente'
DROP_COLS = [ID_COL, 'codigo_admision']

def preprocess(df: pd.DataFrame, reference: pd.DataFrame | None = None) -> pd.DataFrame:
    df = df.drop(columns=[c for c in DROP_COLS if c in df.columns], errors='ignore')
    ref = reference if reference is not None else df
    # Numéricas: imputar con mediana de la referencia
    for col in df.select_dtypes(include=[np.number]).columns:
        df[col] = df[col].fillna(ref[col].median())
    # Categóricas y booleanas: imputar con moda y label-encode
    for col in df.select_dtypes(include=['object', 'bool']).columns:
        if col == TARGET:
            continue
        fill = ref[col].mode()
        df[col] = df[col].fillna(fill.iloc[0] if len(fill) else 'desconocido').infer_objects(copy=False)
        cats = sorted(pd.Series(ref[col].dropna().unique()).astype(str).tolist())
        mapping = {c: i for i, c in enumerate(cats)}
        df[col] = df[col].astype(str).map(mapping).fillna(-1).astype(int)
    return df

y_train = train[TARGET].values

X_train_raw = train.drop(columns=[TARGET])

X_train = preprocess(X_train_raw.copy())
X_test  = preprocess(test.copy(), reference=X_train_raw)

# Alineamos columnas por si en el test faltara alguna categoría del train
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

print('Features del modelo:', X_train.shape[1])
X_train.head(5)
```

## Modelo Base

```python
clf = DecisionTreeClassifier(max_depth=8, min_samples_leaf=20, random_state=42)
clf.fit(X_train, y_train)

print('Accuracy en train:', round(clf.score(X_train, y_train), 3))
print('Features más importantes (top 10):')
imp = pd.Series(clf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
print(imp.head(10).round(3))
```

## Predicción sobre el test

El *predict* devuelve directamente el nombre de la clase (por el modelo que elegí, sino... codificar).

```python
y_pred = clf.predict(X_test)
print('Distribución de predicciones:')
print(pd.Series(y_pred).value_counts(normalize=True).round(3))
```

```python
submit = pd.DataFrame({
    'id_paciente': test['id_paciente'].values,
    'nivel_urgencia': y_pred,
})
assert submit['id_paciente'].is_unique, 'id_paciente duplicado en el submit'
assert submit['nivel_urgencia'].isin(['No urgente', 'Urgente', 'Muy urgente', 'Crítico']).all(), \
    'Hay etiquetas fuera del vocabulario permitido'

out = Path('submission_baseline.csv')
submit.to_csv(out, index=False)
print(f'Guardado: {out.resolve()}  ({len(submit)} filas)')
submit.head()
```

## Métrica del Desafío

Implementación de la métrica, puede ser utilizada en tests internos de modo de evaluar la calidad de la solución con la misma métrica que utiliza el servidor.

```python
LABELS = ["No urgente", "Urgente", "Muy urgente", "Crítico"]

COST_MATRIX = np.array([
    [0, 1, 2, 4],    # Real: No urgente
    [2, 0, 1, 3],    # Real: Urgente
    [4, 2, 0, 2],    # Real: Muy urgente
    [10, 6, 2, 0],   # Real: Crítico
], dtype=float)

_LABEL_TO_IDX = {lbl: i for i, lbl in enumerate(LABELS)}


def hospital_cost_score(y_true, y_pred) -> float:
    """Score = 1 / (1 + costo_promedio). Rango (0, 1]."""
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    if len(y_true) != len(y_pred):
        raise ValueError("y_true e y_pred deben tener la misma longitud")
    idx_true = np.array([_LABEL_TO_IDX[y] for y in y_true])
    idx_pred = np.array([_LABEL_TO_IDX[y] for y in y_pred])
    total_cost = COST_MATRIX[idx_true, idx_pred].sum()
    return float(1.0 / (1.0 + total_cost / len(y_true)))
```

```python
# Supongamos un split de tests con su respectiva predicción
y_test_interno = ["No urgente", "Urgente", "Muy urgente", "Crítico", "Urgente", "Urgente", "Urgente", "Crítico"]
y_pred_interno = ["No urgente", "No urgente", "Muy urgente", "No urgente", "Urgente", "Urgente", "Crítico", "Crítico"]

# La métrica sería:
hospital_cost_score(y_test_interno, y_pred_interno)
```
